```python
import pandas as pd
import numpy as np
```

```python
dataset = {"Name":["Zubaer","Rahim","Karim","Salma"],
 "Age":[22,np.nan, 24, np.nan],
 "Salary":[50000,35000,np.nan, 44000],
 "Gender":['M','M',np.nan,'F']

 }
```

```python
df = pd.DataFrame(dataset)

df
```

# **Missing Value Handling**

```python
df.isnull().sum() #check missing Value
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
df
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
```

```python
df
```

```python
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
```

```python
df
```

# **Group Base Analysis**

```python
sales=pd.DataFrame({
  "City":['Dhaka','Khulna',"Dhaka",'Dhaka','Khulna'],
  "Category" :['Laptop','mouse','mouse','Laptop','Laptop'],
  "Sales":[20000,4000,3000,30000,50000]

})

sales
```

```python
sales.groupby('Category')['Sales'].mean()
```
